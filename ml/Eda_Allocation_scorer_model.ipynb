```python
class Allocation_Scorer:
    def __init__(self, Task, Employees):
        self.Task = Task
        self.Employees = Employees.copy()

    def skill_score(self, emp):
        required = set(self.Task["Skills"])
        available = set(emp["Skills"])
        matched = len(required.intersection(available))

        if matched == 0:
            return 5   # Minimum score floor

        return (matched / len(required)) * 30
        
    def workload_score(self, emp):
          workload = emp["Current_Workload"]
          score = max(0, 25-workload)
          return score   

    def availability_score(self, emp):
            return 20 if emp["Available"] else 0
    def department_score(self, emp):
        if emp["Department"] == self.Task["Department"]:
            return 10
        return 0

    def performance_score(self, emp):
            return emp["Performance"] *10

    def priority_bonus(self):
        priority = self.Task["Priority"]
        if priority == "High":
            return 5
        if priority == "Medium":
            return 3
        return 0
    
    def rank(self):

        scores = []

        bonus = self.priority_bonus()

        for _, emp in self.Employees.iterrows():

            total = (
                self.skill_score(emp)
                + self.workload_score(emp)
                + self.availability_score(emp)
                + self.department_score(emp)
                + self.performance_score(emp)
                + bonus
            )

            scores.append(round(total,2))

        self.Employees["Allocation Score"] = scores

        self.Employees = self.Employees.sort_values(by="Allocation Score", ascending=False)

        return self.Employees
```
